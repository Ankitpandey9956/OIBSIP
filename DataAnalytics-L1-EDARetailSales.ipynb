```python
# plot monthly and quarterly sales trends using line charts
# monrthly Sales
monthly_sales = df.groupby('month')['weekly_sales'].sum()
```

```python
plt.figure(figsize=(7,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks([1,2,3,4], ["Q1","Q2","Q3","Q4"])
plt.grid(True)
plt.show()
```

## observition 
- Weekly sales vary across diffrent months, indicating seasonal trends.
- Some months consistently show higher sales than others.
- peak sales months may correspond to holidays.
- Lower sales months may indicate periods of reduced customer demand.

```python
# quaterly sales.
df['quaterly'] = ((df['month'] - 1) // 3) + 1
```

```python
quaterly_sales = df.groupby('quaterly')['weekly_sales'].sum()
```

```python
plt.figure(figsize=(7,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker='o')
plt.title('Quaterly Sales')
plt.xlabel('Quater')
plt.ylabel('Total Sales')
plt.xticks([1,2,3,4], ["Q1","Q2","Q3","Q4"])
plt.grid(True)
plt.show()
```

## Observation

- Quarterly analysis provides a broader view of sales performance.
- One or more quarters generated significantly higher sales than the others.
- This suggests that business performance changes across seasons.
- The results can help management plan inventory and marketing strategies.

```python
#top 10 best-selling products; revenue by product category (bar chart)
top_ten_sales = df.sort_values(by = 'weekly_sales', ascending = False).head(10)
top_ten_sales
```

```python
plt.figure(figsize=(8,5))

plt.bar(top_ten_sales.index.astype(str),top_ten_sales['weekly_sales'])

plt.title("top ten weekly sales")
plt.xlabel('Record Index')
plt.ylabel('Weekly Sales')

plt.show()

```

## Observation

- The chart highlights the ten highest weekly sales records.
- These sales may have occurred during promotional events or peak shopping seasons.
- There is a noticeable difference between the highest-selling weeks and the remaining weeks.
- Identifying these periods can help understand what drives exceptional sales performance.

```python
numberic_df = df.select_dtypes(include='number')

```

```python
corr_matrix = numberic_df.corr()
```

```python
plt.figure(figsize =(10,8))

sns.heatmap(
    corr_matrix,annot=True,cmap='coolwarm',fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()
```

## Observation

- The heatmap shows the correlation between numerical variables.
- Weekly sales are strongly related to units sold, which is expected.
- Promotion discounts may show a positive relationship with sales, indicating that promotions can increase revenue.
- Some variables have weak or no correlation, suggesting they have little linear relationship.

```python
#  At least one additional visualisation of your choice that reveals a non-obvious insight
plt.figure(figsize=(8,6))
plt.scatter(
    df['promotion_discount'],
    df['weekly_sales'],
    alpha=0.6
)

plt.title('Promotion Discount vs Weekly Sales')
plt.xlabel('promotion Discount')
plt.ylabel('Weekly Sales')

plt.grid(True)

plt.show()
```

## Observation

- Weekly sales during promotional periods appear higher than during non-promotional periods.
- Promotions also introduce greater variation in sales.
- A few weeks achieved exceptionally high sales, shown as outliers.
- This suggests that promotional campaigns positively influence customer purchases.
